## Prompt Template

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template = "{instructions} my name is {name}",
    variale_name = ["name"],
    partial_variable = {"instructions": instruction})
```

```python
prompt.invoke({"name": "Alice", "instructions": "Answer the following question truthfully." })
```

## ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = .from_messages([
    SystemMessage(content="You are a helpful assistant."),  
    HumanMessage(content="Hello, who won the world series in 2020?"),
    AIMessage(content="TChatPromptTemplatehe Los Angeles Dodgers won the World Series in 2020."),
    HumanMessage(content="Where was it {played}?")
])
```

```python
prompt.invoke({"played": "held" })
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate(
    messages = [
        MessagesPlaceholder(variable_name="history"),
        ("human","Tell me a joke on {topic} .")
    ]
)
```

```python
hist =  [AIMessage(content="Why did the scarecrow win an award? Because he was outstanding in his field!")]
res = prompt.invoke({"topic":"ipl","history":hist})
print(res)
```
